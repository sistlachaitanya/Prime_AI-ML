```python
import json
```

```python
def load_data(file_name):
    with open(file_name, "r") as f:
        data = json.load(f)
    return data
```

```python
data = load_data("store_data.json")
print(data)
print(type(data))
```

```python
def clean_data(data):
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_users = set()
    for user in data:
        raw_rating = str(user["rating"]).strip().lower()
        if raw_rating in text_to_num:    
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating  

        raw_age = user.get("age")
        if not raw_age:
            user["age"] = None
            
        if user["name"] in unique_users:
            continue
        unique_users.add(user["name"])
        cleaned_data.append(user)
    return cleaned_data
```

```python
data = clean_data(data)
```

```python
def get_insights(data):
    tot_rating = 0
    poor_ratings = 0

    for user in data:
        tot_rating += float(user["rating"])
        if float(user["rating"]) < 3:
            poor_ratings += 1 
        
    print(f"average rating = {tot_rating/len(data)}")
    print(f"poor ratings = {poor_ratings}")
    
```

```python
get_insights(data)
```

```python
def get_recommendations(data):
    recommandations = []

    for user in data:
        curr_recomm = {}
        curr_recomm["user"] = user["name"]

        if float(user["rating"]) > 3:
            curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["brand"] = "Samsung"
        recommandations.append(curr_recomm)
    return recommandations
```

```python
get_recommendations(data)
```
