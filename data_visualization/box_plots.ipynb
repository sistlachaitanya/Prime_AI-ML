```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Box plot

data = [-5, 7, 8, 5, 6, 9, 4, 10, 12, 15, 20] 
plt.boxplot(data)
plt.grid()
```

```python
Q1 = np.percentile(data, 25)
Q2 = np.median(data) 
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

print("Lower Whisker:", Q1-(1.5*IQR))
print("Q1, P25:", Q1)
print("Q2 Median P50:", Q2)
print("Q3, P75:", Q3)
print("Upper Whisker:", Q3+(1.5*IQR))
```

```python
group1 = np.random.normal(50, 10, 100)
group2 = np.random.normal(60, 15, 100)

plt.boxplot([group1, group2])
```

```python
plt.boxplot([group1, group2], tick_labels=["group1", "group2"], vert=False, showmeans=True)
```
