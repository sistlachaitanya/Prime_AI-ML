```python
import matplotlib.pyplot as plt
```

```python
# Pie Charts
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]
```

```python
explode = [0, 0, 0.1, 0, 0]

plt.pie(amounts, labels=expenses, autopct="%1.0f%%", explode=explode, shadow=True)
```
