```python
import matplotlib.pyplot as plt
```

```python
# Scatter Plots

people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]

colors=["red", "blue", "green", "yellow", 
        "purple", "pink", "black", "skyblue", 
        "red", "grey"]

bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
```

```python
plt.scatter(age, bp, color=colors)
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

plt.grid()
```

```python
colors = ["green" if x < 135 else "red" for x in bp]

plt.scatter(age, bp, color=colors)
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

plt.grid()
```

```python
plt.scatter(age, bp, s=bp, cmap="OrRd", c=bp)
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

plt.grid()
plt.colorbar()

for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i], bp[i]), xytext=(age[i], bp[i]))
```
