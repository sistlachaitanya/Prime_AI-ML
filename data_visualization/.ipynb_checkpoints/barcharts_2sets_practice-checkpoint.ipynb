```python
import matplotlib.pyplot as plt
import numpy as np

months = ["Jan", "Feb", "Mar", "Apr", "May"]

company_A = [40, 50, 60, 70, 80]
company_B = [30, 60, 50, 90, 70]

x = np.arange(len(months))
width = 0.35

plt.bar(x - width/2, company_A, width, label="Company A")
plt.bar(x + width/2, company_B, width, label="Company B")

plt.xticks(x, months)

plt.legend()
plt.show()
```
