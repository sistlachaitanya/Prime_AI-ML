```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]
steps = [3000, 4500, 4000, 6000, 8000, 7500, 9000]

# 1. Explore available styles
print(plt.style.available)

# 2. Apply a style
plt.style.use("ggplot")

# Create the plot
plt.plot(days, steps, marker="o", label="Steps")

# Add title and labels
plt.title("Weekly Step Count")
plt.xlabel("Day")
plt.ylabel("Number of Steps")

# 3. Add grid lines
plt.grid(True, linestyle="--", alpha=0.5)

plt.legend()

# 4. Adjust spacing
plt.tight_layout()

# 5. Save the figure
plt.savefig("weekly_steps.png", dpi=300, bbox_inches="tight")

plt.show()
```
