```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

steps = [3000, 4500, 4000, 6000, 7500, 9000, 8000]
calories = [200, 300, 250, 400, 500, 600, 550]

plt.plot(days, steps, "r.--", label="Steps")
plt.plot(days, calories, marker="s", label="Calories")

plt.title("Weekly Activity")
plt.xlabel("Days")
plt.ylabel("Values")
plt.legend()

plt.show()
```
