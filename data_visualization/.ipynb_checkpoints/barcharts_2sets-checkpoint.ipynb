```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

non_oscar_movies = [
    "Slumdog Millionaire",
    "Avatar",
    "Inception",
    "Hugo",
    "Lincoln"
]

non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M
years = [2008, 2009, 2010, 2011, 2012]
```

```python
x = np.arange(len(years))
width = 0.35

plt.bar(x-width/2, oscar_revenue, width, label="Oscar Movies")
plt.bar(x+width/2, non_oscar_revenue, width, label="Nn Oscar Movies")

plt.title("Oscar vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue in ($M)")
plt.legend()
plt.xticks(x, years)
```

```python
plt.barh(oscar_movies, oscar_revenue)
plt.xlabel("Revenue in ($M)")
plt.ylabel("Movies")
plt.title("Revenue by Oscar Movies")
```
