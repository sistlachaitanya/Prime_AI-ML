```python
import matplotlib.pyplot as plt
```

```python
years = [2024, 2025, 2026]

development = [20, 30, 40]

testing = [10, 15, 20]

hr = [5, 10, 15]
```

```python
plt.plot(years, development, label="dev employees", marker="o")
plt.plot(years, testing, label="testing employees", marker="^")
plt.plot(years, hr, label="hr employees", marker="o")
plt.tight_layout()
plt.legend()
```

```python
import matplotlib.pyplot as plt

years = [2023, 2024, 2025, 2026]

development = [20, 30, 40, 50]
testing = [10, 15, 20, 25]
hr = [5, 10, 15, 20]

plt.stackplot(
    years,
    development,
    testing,
    hr,
    labels=["Development", "Testing", "HR"],
    colors=["#2ECC71", "#3498DB", "#F39C12"]
)

plt.legend()
plt.title("Department-wise Employee Growth")
plt.xlabel("Year")
plt.ylabel("Number of Employees")

plt.show()
```
