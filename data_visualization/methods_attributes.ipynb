```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

steps = [3000, 4500, 4000, 6000, 7500, 9000, 8000]
calories = [200, 300, 250, 400, 500, 600, 550]

plt.style.use("dark_background")
plt.grid()

plt.plot(days, steps)

plt.title("Weekly Activity")
plt.xlabel("Days")
plt.ylabel("Steps")

plt.show()
```

```python
plt.plot(days, steps, marker="o")

plt.title("Daily Steps")
plt.xlabel("Day of the Week")
plt.ylabel("Number of Steps")

plt.tight_layout()
plt.show()
```
