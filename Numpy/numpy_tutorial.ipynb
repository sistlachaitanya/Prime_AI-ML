```python
import numpy as np
import time
```

```python
size = 1_000_000

#Python List
py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"Python List time = {end - start} seconds")

# Numpy array
np_arr = np.array(py_list)
start = time.time()
sq_arr = np_arr**2
end = time.time()
print(f"Numpy array time = {end - start} seconds")
```

```python
import sys
print(f"python list size = {sys.getsizeof(py_list) * 4}")
print(f"numpy array size = {np_arr.nbytes}")
```
